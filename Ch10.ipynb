- 장기 의존성 문제
    - 시퀀스의 길이가 길어질수록 초기에 처리한 정보에 대한 기억력 감소 발생

    - 텍스트 시퀀스를 입력받아 다른 도메인의 텍스트 시퀀스를 출력하는 구조 채택
    - 디코더가 인코더의 마지막 은닉 상태만을 참조하여 번역을 수행하는 정보 병목 현상 존재

- 자기 회귀 속성
    
    디코더가 이전에 생성된 토큰을 참조하여 다음 토큰을 예측 및 생성하는 방식

##### **어텐션 메커니즘**

<img src='https://github.com/HANISY/AI-Study/blob/main/img/attention%20mechanism.png?raw=true'>

- 기존 모델의 한계 극복 및 성능 향상
    - 순환 신경망(RNN) 기반 인코더-디코더 모델의 번역 성능을 비약적으로 개선한 기술
    - 인코더의 마지막 은닉 상태에만 의존하던 기존 방식 탈피
    - 인코더의 모든 타임스텝에서 계산된 은닉 상태를 디코더가 참조 가능
- 어텐션 가중치
    - 디코더가 입력 토큰마다 중요도를 다르게 부여하는 방식 적용
    - 모든 은닉 상태를 동일하게 참조하지 않고, 각 상태마다 상이한 가중치를 적용하여 핵심 정보 강조
- 어텐션 메커니즘의 장점과 단점
    - 장점: 정보 손실 완화
        - 시퀀스 길이가 긴 텍스트 처리 시 발생하는 정보 손실 문제의 효과적 해결
    - 단점: 연산 비용 및 길이 제한
        - 참조해야 할 정보량 증가에 따른 연산량 상승
        - 인코더가 처리 가능한 최대 타임스텝 개수 지정 필요(입력 텍스트 길이의 제한 존재)

##### **트랜스포머**

<img src='https://github.com/HANISY/AI-Study/blob/main/img/transformer.png?raw=true'>

- 기본 구조 및 특징
    - 어텐션 메커니즘을 핵심으로 다양한 기술을 조합하여 구성
    - 기존의 인코더-디코더 프레임워크는 유지하되, 순환 신경망(RNN)을 완전히 제거한 구조
- 인코더의 병렬 처리 방식
    - 입력된 텍스트 전체를 한 번에 처리하는 병렬 연산 수행
    - 순차적 처리가 아니므로 타임스텝( 개념이 불필요하며 처리 속도가 비약적으로 향상됨
    - 어텐션 메커니즘 특성상 입력 텍스트 길이에 대한 제한은 존재
- 데이터 표현 용어의 변경
    - RNN 미사용으로 인한 '은닉 상태' 개념의 소멸

    - 인코더로부터 전달받은 은닉 벡터를 활용하여 각 단계의 출력 토큰 생성
    - 이전에 생성된 토큰을 참조하여 다음 토큰을 예측하는 자기회귀 속성 유지
    - 순환 신경망 없이도 이전 출력 정보를 반영할 수 있는 구조 확보

##### **셀프 어텐션 메커니즘**

- 셀프 어텐션의 정의 및 학습 목표
    - 기존 어텐션과 달리 인코더 내부의 입력 토큰들 간의 상호 관계를 학습하는 메커니즘
    - 각 토큰의 벡터 표현이 주어진 문제 해결에 최적화되도록 쿼리, 키, 값 생성용 가중치를 학습
- 셀프 어텐션 연산 프로세스
    - 벡터 생성: 입력 텍스트를 서로 다른 가중치를 가진 세 개의 밀집층에 통과시켜 쿼리, 키, 값 벡터를 각각 생성
    - 어텐션 점수 계산: 쿼리 벡터와 키 벡터의 곱을 통해 토큰 간 연관성을 수치화한 어텐션 행렬 산출
    - 최종 출력 생성: 계산된 어텐션 점수를 값 벡터에 곱하여 중요도가 반영된 최종 은닉 벡터 도출
- 문맥 이해와 정보 강조
    - 출력 벡터는 해당 토큰이 문장 내 다른 토큰들과 얼마나 밀접하게 관련되어 있는지를 반영
    - 이를 통해 모델이 문맥을 정교하게 파악하고 핵심 정보를 효과적으로 강조함
- **멀티 헤드 어텐션**
    - 어텐션 헤드: 셀프 어텐션 연산을 수행하는 독립적인 연산 단위
    - 다양성 확보: 각 헤드마다 Q, K, V를 생성하는 밀집층을 다르게 구성하여 동일한 문장을 다양한 관점에서 해석
    - 결합: 여러 헤드의 출력 벡터를 하나로 연결한 후, 최종 밀집층을 통과시켜 어텐션의 최종 결과물 완성

##### **층 정규화**

딥러닝에서는 여러 개의 층을 거치면서 특성의 스케일이 변할 수 있기 때문에 단순환 입력 정규화만으로 충분하지 않음

- 배치 정규화의 원리 및 한계
    - 합성곱 신경망(CNN)의 층 사이에서 배치 단위의 평균과 분산을 이용한 정규화 수행
    - 데이터 분포를 평균 0, 분산 1로 조정하되 정보 손실 방지를 위한 학습 가능한 두 개의 파라미터(Scale, Shift) 적용
    - 훈련 속도 가속 및 학습 안정화 효과 존재하나 가변적인 길이를 가진 텍스트 데이터에 대한 적용의 어려움
- 층 정규화의 특징
    - 각 샘플의 토큰마다 독립적인 평균과 분산을 계산하여 정규화 적용
    - 샘플 간 길이 차이와 무관하게 독립적 연산이 가능하여 트랜스포머 등 NLP 모델에 적합
    - 멀티 헤드 어텐션 층 이후 드롭아웃과 함께 적용되는 구조

<img src='https://github.com/HANISY/AI-Study/blob/main/img/layer%20nomalization.png?raw=true'>

- 잔차 연결의 메커니즘 및 효과
    - ResNet에서 최초 도입된 기술로 입력값을 연산 출력값에 그대로 더해주는 스킵 연결 방식
    - 멀티 헤드 어텐션 층의 출력에 입력을 더하여 정보 흐름 개선
    - 역전파 시 그래디언트(업데이트 신호)가 복잡한 층을 건너뛰어 앞쪽 층으로 직접 전달됨에 따라 깊은 신경망의 효과적인 학습 지원
    - 트랜스포머 인코더 내 두 개의 잔차 연결 구조 포함

##### **피드포워드 네트워크와 인코더 블록**

**피드포워드 네트워크의 구조**

    - 첫 번째 밀집층: ReLU 활성화 함수 적용
    - 두 번째 밀집층: 활성화 함수 미적용(Linear)
- 후처리 및 연결 방식
    - 두 번째 밀집층 이후 드롭아웃 층 추가
    - 전체 피드포워드 네트워크를 감싸는 잔차 연결 적용

- 인코더 블록 구성 요소

    - 블록의 최종 단계에 층 정규화를 배치하여 하나의 온전한 인코더 블록 형성
- 입출력 차원의 일치 및 확장성

    - 출력: 각 토큰에 대한 은닉 벡터
    - 차원 유지: 입력되는 벡터의 차원과 블록이 출력하는 은닉 벡터의 차원이 동일함
- 블록 적층을 통한 모델 구성
    - 입출력 차원이 동일하므로 같은 구조의 인코더 블록을 반복적으로 쌓는 것이 가능
    - 모델의 복잡도와 성능 요구사항에 따라 수 개에서 수십 개의 블록을 순차적으로 연결하여 최종 인코더 모델 형성

##### **디코더 블록**

- 크로스 어텐션 메커니즘
    - 인코더와 디코더의 상호작용을 담당하며 인코더의 정보를 디코더로 통합하는 계층
    - 쿼리: 디코더의 이전 층에서 넘어온 벡터 사용
    - 키 & 값: 인코더의 최종 출력 은닉 벡터 사용
    - 디코더 블록 내부의 '마스크드 멀티 헤드 어텐션'과 '피드포워드 네트워크' 사이에 배치됨
- 디코더 블록의 적층(Stacking) 및 정보 전달
    - 단일 블록이 아닌 다수의 디코더 블록을 반복적으로 쌓아 전체 모델을 구성
    - 인코더의 출력 정보는 첫 번째 디코더 블록뿐만 아니라 모든 디코더 블록에 동일하게 전달되어 활용됨
    - 각 블록은 입출력 크기가 동일한 은닉 벡터를 생성

<img src='https://github.com/HANISY/AI-Study/blob/main/img/transformer%20decoder%20block.png?raw=true'>

- 최종 출력층 구성
    - 반복된 디코더 블록의 마지막 단계 이후, 수행하려는 작업에 적합한 층을 배치
    - 텍스트 분류나 번역의 경우, 주로 밀집층을 추가하여 다음 토큰이나 클래스를 예측하도록 설계
- 마스크드 멀티 헤드 어텐션
    - 디코더 블록의 최상단(맨 왼쪽)에 위치하며 인코더와 가장 차별화되는 부분
    - 자기회귀 속성: 한 번에 하나의 토큰만 생성하는 순차적 방식 준수
    - 마스킹 적용: 현재 타임스텝 이후의 미래 토큰(정답)을 참조할 수 없도록 가림 처리
    - 이를 통해 모델이 정답을 단순 복사하지 않고 문맥을 학습하도록 유도

**대규모(Large Language Model, LLM)의 개념**
- 트랜스포머(Transformer) 아키텍처를 기반으로 설계된 거대 신경망 모델
- 수억 개에서 수천억 개에 이르는 방대한 양의 파라미터를 보유
- 대규모 데이터 학습
    - 인터넷상의 방대한 텍스트 데이터셋(말뭉치)을 통해 언어의 패턴과 지식을 학습
    - 문맥 이해, 생성, 요약 등 다양한 자연어 처리(NLP) 작업 수행 가능
- 자원 및 비용의 문제
    - 컴퓨팅 자원: 모델 훈련을 위해 수천 개의 고성능 GPU/TPU 등 막대한 하드웨어 인프라 필요
    - 비용 소모: 방대한 데이터 처리와 긴 학습 시간으로 인해 천문학적인 전력 및 운용 비용 발생
    - 이러한 진입 장벽으로 인해 주로 거대 기술 기업이나 연구소 주도로 개발이 이루어짐

##### **트랜스포머 가계도**

**인코더 및 시퀀스-투-시퀀스 모델의 주요 응용**
- 개체명 인식
    - 텍스트에서 인명, 지명, 기관명 등 고유 명사를 식별 및 분류하는 작업
    - 인코더를 활용하여 입력된 각 토큰별로 개체명 여부 및 범주를 출력
- 의미론적 텍스트 유사도(STS, Semantic Textual Similarity)
    - 두 개의 텍스트가 의미적으로 얼마나 유사한지를 수치화하여 측정하는 작업
    - 인코더를 통해 추출된 문맥 벡터 간의 비교 연산 수행
- 질문-답변(QA, Question-Answering)
    - 질문과 문맥이 주어졌을 때 적절한 응답을 생성하는 대표적인 시퀀스-투-시퀀스 작업
    - 문맥 내에서 정답을 탐색하거나 생성하여 답변 도출

**대규모 언어 모델(LLM)의 유형 및 발전**
- 초기 모델의 한계 및 진화
    - 초창기에는 트랜스포머 구조 없이 단어 임베딩 벡터 생성에 주력
    - 이후 트랜스포머 아키텍처 도입으로 비약적인 성능 향상 달성
- 아키텍처별 대표 모델 분류
    - 인코더 기반: BERT, RoBERTa, ELECTRA (이해 및 분류 작업에 강점)
    - 인코더-디코더 기반: T5, BART (번역, 요약 등 변환 작업에 강점)
    - 디코더 기반: GPT-3, ChatGPT (자연스러운 텍스트 생성에 강점)

**디코더 기반 모델의 인기 요인과 프롬프트**
- 텍스트 생성 능력의 범용성
    챗봇, 작문, 번역, 요약 등 다양한 애플리케이션에 유연하게 적용 가능한 생성 능력 보유
- 자기회귀적 작동 원리
    - 무에서 유를 창조하는 것이 아닌, 이전 텍스트를 기반으로 다음 토큰을 예측하는 메커니즘
    - 이전에 생성된(혹은 입력된) 텍스트를 문맥으로 삼아 연쇄적인 토큰 생성 수행
- 프롬프트의 역할
    - 모델에게 입력하는 지시문이나 예시 텍스트
    - 디코더가 마치 이전에 생성한 텍스트인 것처럼 인식하게 하여, 인코더 없이도 다음 내용을 예측하고 작업을 수행하도록 유도하는 기술

- 2019년 메타에서 공개한 트랜스포머 기반의 시퀀스-투-시퀀스(인코더-디코더) 언어 모델
- 약 160GB의 대규모 텍스트 데이터셋을 사용하여 사전 훈련 수행
- 베이스 모델 구성
    - 인코더 블록과 디코더 블록을 각각 6개씩 적층하여 구성 (Large 모델의 경우 12개씩 사용)
    - 블록(Layer)의 개수가 증가할수록 모델의 파라미터 수가 비례하여 증가하는 구조

<img src='https://github.com/HANISY/AI-Study/blob/main/img/BART.png?raw=true'>

**임베딩 및 위치 인코딩 방식**
- 학습 가능한 위치 임베딩
    - 기존 트랜스포머의 고정된 사인/코사인 함수 대신, 정수를 실수 벡터로 변환하는 임베딩 층을 위치 정보 표현에 사용
    - 토큰 임베딩과 마찬가지로 초기에는 무작위로 초기화되며, 훈련 과정을 통해 최적의 위치 값을 학습함
- 공유 임베딩 층
    - 인코더와 디코더에 입력되는 토큰은 모두 동일한 하나의 임베딩 층을 통과
    - 단어의 의미를 일관되게 해석하고 엉뚱한 결과 출력을 방지하기 위해 인코더와 디코더가 같은 임베딩 파라미터를 공유
- 입력 처리 흐름
    - 토큰 임베딩 벡터와 위치 임베딩 벡터를 합산
    - 층 정규화 및 드롭아웃 수행 후 인코더/디코더 블록으로 전달

**출력층의 가중치 공유 및 효율성**
- 별도 밀집층 제거를 통한 파라미터 절감
    - 디코더의 마지막 출력을 단어 확률로 변환할 때, 새로운 밀집층을 두지 않고 입력 임베딩 층의 행렬을 재사용
    - 모델 전체의 파라미터 수를 줄이고 학습 효율성을 높이는 기법
- 출력 확률 계산 메커니즘입력 임베딩 행렬을 전치(Transpose, 세로로 회전)시킨 후 디코더의 마지막 출력 벡터와 곱셈 연산 수행
- 연산 과정:
    - 디코더 출력 벡터 $\times$ 전치된 임베딩 행렬의 첫 번째 열 = 첫 번째 토큰이 선택될 점수(Logit)
    - 디코더 출력 벡터 $\times$ 전치된 임베딩 행렬의 두 번째 열 = 두 번째 토큰이 선택될 점수
    - 계산된 점수를 소프트맥스 함수 등을 통해 최종 확률값으로 변환하여 다음 토큰 예측

##### **BART의 인코더와 디코더**

<img src='https://github.com/HANISY/AI-Study/blob/main/img/BART2.png?raw=true'>

- 활성화 함수: 젤루(GeLU) 도입
    -  원본 트랜스포머가 렐루(ReLU)를 사용하는 것과 달리, BART는 젤루(GeLU, Gaussian Error Linear Unit) 함수를 채택
    - **GeLU**의 정의: 입력값에 표준 정규 분포의 누적 분포 함수(CDF)를 곱하는 방식 ($x \times \Phi(x)$)
    - 음수 입력에서도 미세한 기울기를 허용하여 0으로 완전히 죽는 현상을 방지하고, 곡선형의 부드러운 변화를 유도
- 피드포워드 네트워크의 차원 변화
    - 첫 번째 밀집층: 입력 임베딩 벡터 차원의 4배로 확장 (대부분의 트랜스포머 모델에서 통용되는 비율이나 절대적인 규칙은 아님)
    - 두 번째 밀집층: 확장된 차원을 다시 원래의 임베딩 차원으로 축소
    - 이 과정을 통해 차원은 유지하면서 정보의 특징을 비선형적으로 변환
- 드롭아웃과 최신 LLM 트렌드
    - BART는 인코더와 디코더의 주요 연결 지점에 비율 0.1의 드롭아웃 층을 적용하여 과대적합 방지
    - 최신 트렌드: 최근의 초거대 언어 모델(LLM)은 방대한 데이터셋 크기와 학습 비용 문제로 인해 1 에포크 이상 훈련하기 어려움. 따라서 과대적합 위험이 상대적으로 낮아 드롭아웃을 사용하지 않거나 비율을 낮추는 추세

**인코더 블록의 데이터 처리 흐름**

1\) 멀티 헤드 어텐션: 입력된 임베딩 벡터를 여러 헤드로 나누어 처리 후, 밀집층을 통해 다시 입력 크기의 벡터로 병합<br>
2\) 1차 후처리: 드롭아웃 $\rightarrow$ 잔차 연결 $\rightarrow$ 층 정규화 수행<br>
3\) 피드포워드 네트워크: 확장(4배) 및 축소(1배) 과정을 거치며 특징 추출<br>
4\) 2차 후처리: 드롭아웃 $\rightarrow$ 잔차 연결 $\rightarrow$ 층 정규화 수행<br>
5\) 최종 출력: 입력과 동일한 크기의 벡터를 출력하여 다음 블록으로 전달<br>

**디코더 블록의 데이터 처리 흐름**
- 구조적 유사성과 차이점<br>
    인코더와 유사한 처리 과정을 거치나, 두 개의 어텐션 층을 포함한다는 점이 다름
- 단계별 처리 과정
    - 마스크드 멀티 헤드 어텐션: 12개의 헤드로 분할 처리. 현재 시점 이후의 정보를 가리는 마스킹 적용 후 밀집층 통과. 이어지는 드롭아웃, 잔차 연결, 층 정규화 수행
    - 크로스 어텐션
        - 쿼리: 디코더의 이전 단계 출력 벡터
        - 키 & 값: 인코더의 최종 출력 벡터
        - 인코더 정보를 디코더 흐름에 통합한 후 드롭아웃, 잔차 연결, 층 정규화 수행
    - 피드포워드 네트워크 & 후처리: 인코더와 동일한 구조(4배 확장 후 축소)를 거친 뒤 최종 정규화 수행

##### **텍스트 토큰화**

**토큰화와 토크나이저의 역할**
- 토큰화의 정의 및 입력 처리
    - 텍스트 데이터를 모델이 처리할 수 있는 기본 단위인 '토큰'으로 분할하는 과정
    - 트랜스포머 모델은 원시 텍스트가 아닌, 각 토큰에 매핑된 정수 ID(Integer ID)의 리스트를 입력으로 수신한다고 가정
- 토크나이저의 기능 및 학습
    - 훈련 데이터로부터 데이터셋의 특성을 반영한 최적의 어휘 사전을 구축(학습)하는 도구
    - 텍스트를 정수로 변환(Encoding)하거나 정수를 다시 텍스트로 복원(Decoding)하는 역할 수행
- 토크나이저와 임베딩 층의 구분
    - 토크나이저: 텍스트와 정수 ID 간의 매핑 규칙(사전)만 관리
    - 임베딩 층: 정수 ID를 실수 벡터로 변환하는 **임베딩 벡터는 토크나이저가 아닌 LLM 모델 내부의 첫 번째 층(Embedding Layer)에 존재**하며 모델 학습 시 업데이트됨

**공백 분할의 한계와 부분단어 토큰화**
- 공백 기반 분할의 문제점
    - 단어를 공백 기준으로만 나눌 경우, 훈련 데이터에 등장하지 않은 단어(Unknown Token, OOV)는 처리가 불가능하거나 별도의 토큰(<UNK>)으로 처리되어 정보 손실 발생
    - 어휘 사전의 크기가 불필요하게 커지는 희소성 문제 유발
- 부분단어 토큰화의 도입
    - 단어보다 더 작은 단위로 텍스트를 분해하는 방식
    - 빈번하게 등장하는 단어는 하나의 토큰으로, 희귀한 단어는 여러 개의 부분단어 조합으로 표현
    - 처음 보는 단어라도 알고 있는 부분단어들의 조합으로 인식하여 의미 추론 가능
    - 트랜스포머 기반 최신 LLM의 표준 토큰화 방식으로 자리 잡음
    - BPE, 워드피스, 유니그램, 센텐스피스 등

**BPE(Byte Pair Encoding)**

- 빈도 기반의 병합 알고리즘
    - 초기에는 단어를 문자 단위로 분해하여 어휘 사전을 구성함
    - 전체 데이터셋에서 가장 빈번하게 등장하는 문자 쌍을 찾아 하나의 새로운 부분단어로 병합함
    - 사용자가 지정한 어휘 사전 크기에 도달할 때까지 이 병합 과정을 반복적으로 수행하여 최종 토큰 리스트를 확정함
- 유니코드(Unicode) 환경에서의 구조적 한계
    - 전 세계 언어와 특수문자를 포함하는 유니코드는 기본 문자의 개수가 14만 개 이상으로 매우 방대함
    - 이를 문자 단위로 처리할 경우 기본 어휘 사전의 크기가 지나치게 비대해지는 문제 발생
    - 어휘 사전의 크기 증가는 임베딩 층의 행렬 크기를 키워 모델의 총 파라미터 수를 급격히 증가시키는 원인이 됨
- **바이트 수준 BPE(Byte-level BPE)**
    - 바이트 스트림 처리 방식
        - 텍스트를 유니코드 문자가 아닌 UTF-8 인코딩에 기반한 바이트 시퀀스로 변환하여 인식함
        - 초기 기본 어휘 사전의 크기가 바이트의 표현 범위인 256개로 고정되므로 매우 작게 시작 가능함
    - 유니코드 처리의 효율성 및 범용성
        - 자주 등장하는 바이트 쌍을 병합하여 토큰을 생성하므로, 어떤 언어나 문자(이모티콘 포함)가 입력되더라도 바이트 조합으로 표현 가능함
        - 적절한 크기(예: 5만~6만 개)의 어휘 사전만으로도 미등록 토큰(UNK) 없이 모든 텍스트를 처리할 수 있는 유연성을 확보함
        - GPT-2, GPT-3, RoBERTa 등 현대의 주요 LLM들이 채택하고 있는 표준적인 토큰화 방식임

**워드피스**

- BPE와 유사하게 부분단어 쌍을 병합하지만, 단순히 빈도가 높은 쌍을 선택하지 않음
- 병합하려는 쌍의 빈도뿐만 아니라, 그 쌍을 구성하는 개별 토큰의 빈도까지 고려하여 점수를 산출
- 두 토큰이 독립적으로 등장할 확률 대비 함께 등장할 확률이 높을수록(즉, 결합의 정당성이 높을수록) 우선적으로 병합
- 자주 등장하지 않는 단어라도 의미 있는 단위(어근, 접사 등)로 분리하는 데 효과적
- BERT와 같은 인코더 기반 모델의 표준 토큰화 방식으로 채택됨

**유니그램**

- 초기에 매우 방대한 크기의 어휘 사전을 구축한 뒤, 목표 크기에 도달할 때까지 불필요한 토큰을 제거해 나가는 방식
- BPE나 워드피스가 작은 단위에서 합쳐가는 것과 반대되는 접근법
- 각 토큰이 독립적으로 발생한다고 가정하고, 전체 텍스트의 손실(음의 로그 가능도)을 최소화하는 방향으로 어휘 사전을 최적화
- 특정 토큰을 제거했을 때 전체 손실이 가장 적게 증가하는(가장 덜 중요한) 토큰부터 순차적으로 삭제

- **확률(Probability)**
    - 관점: 모델 파라미터가 고정된 상태에서 새로운 데이터(샘플)를 바라보는 관점
    - 역할: 이미 훈련된 모델이 입력 데이터를 받아 결과를 **예측(Prediction)** 할 때 사용되는 값
- **가능도(Likelihood)**
    - 관점: 데이터셋이 고정된 상태에서 모델의 파라미터를 바라보는 관점
    - 역할: 모델을 **훈련(Training)** 할 때, 현재의 파라미터가 주어진 데이터를 얼마나 잘 설명하는지 평가하는 척도

**센텐스피스**

- 띄어쓰기가 없거나 모호한 언어(중국어, 일본어 등)에서 발생하는 공백 기반 사전 토큰화의 한계 극복
- 입력 텍스트를 단어 단위로 구분하지 않고, 문장 전체를 하나의 문자 스트림(Raw Stream)으로 취급
- 공백(Space) 또한 하나의 문자로 간주하여 특수 기호(예: _)로 치환해 처리
- 내부적으로는 BPE나 유니그램 알고리즘을 선택하여 어휘 사전을 구성하는 라이브러리이자 기법

1\) 입력 및 임베딩

2\) 디코더 블록 (총 30개 적층)

- RMS 정규화: 블록 진입 시 가장 먼저 수행
- 그룹 쿼리 어텐션 (GQA):
    - 입력 벡터(2,560)를 80차원씩 나누어 32개의 쿼리 헤드 생성
    - 키와 값 헤드는 8개만 사용하여 그룹 공유 (GQA 적용)
    - 어텐션 출력은 다시 2,560차원으로 복원 후 잔차 연결 및 2차 RMS 정규화 수행
- 피드포워드 네트워크
    - 두 개의 밀집층을 통해 은닉 벡터 크기를 7,168로 확장 (이 중 하나에만 SiLU 적용 후 곱셈)
    - 마지막 밀집층에서 다시 2,560차원으로 축소
3\) 최종 출력
- 30개의 블록 통과 후 마지막 RMS 정규화 수행
- 별도의 밀집층(Output Head)을 통과하여 어휘 사전 크기인 102,400 차원의 벡터 출력 (Weight Tying을 사용하지 않음)
- 소프트맥스를 통해 최종 확률 분포 산출

**그룹 쿼리 어텐션 (GQA, Grouped Query Attention)**
- 도입 배경: KV 캐시(Key-Value Cache) 문제
    - 자기회귀 모델은 생성 속도 향상을 위해 이전 토큰의 키와 값을 캐시에 저장
    - 입력 길이가 길어질수록 캐시 메모리 사용량이 급증하는 문제 발생
- GQA의 개념 및 효과
    - 멀티 헤드 어텐션(MHA): 모든 쿼리 헤드가 각자의 키/값 헤드를 가짐 (파라미터 많음)
    - 멀티 쿼리 어텐션(MQA): 모든 쿼리 헤드가 하나의 키/값 헤드를 공유 (파라미터 최소화, 성능 저하 우려)
    - 그룹 쿼리 어텐션(GQA): MHA와 MQA의 절충안. 몇 개의 쿼리 헤드를 그룹으로 묶어 키/값 헤드를 공유
    - 키/값 생성을 위한 밀집층 개수를 줄여 파라미터를 절감하고 추론 속도 및 메모리 효율성 확보

<img src='https://github.com/HANISY/AI-Study/blob/main/img/query%20attention.png?raw=true'>

**활성화 함수: SiLU (Sigmoid Linear Unit)**
- 특징 및 동작
    - 입력값에 시그모이드 함수를 적용한 뒤 다시 입력값을 곱하는 형태 ($x \times \sigma(x)$)
    - ReLU와 유사하나 곡선 형태를 띠며, GeLU처럼 0 근처에서도 미분 가능하여 학습 안정성 우수
- 피드포워드 네트워크(FFN) 적용 방식 (Gated Linear Unit 변형)
    - 첫 번째 밀집층을 두 개의 경로로 분할
    - 한쪽은 SiLU 적용, 다른 한쪽은 그대로 둔 후 두 결과를 곱하는 방식(SwiGLU 등과 유사) 채택

**정규화: RMS 정규화 (RMSNorm)**
- 기존 층 정규화와 달리 평균을 계산하거나 빼지 않음

<img src='https://github.com/HANISY/AI-Study/blob/main/img/RMS.png?raw=true'>

- 표준 편차 계산 시에도 평균을 사용하지 않고 제곱 평균 제곱근(RMS)만 사용하여 정규화 수행

<img src='https://github.com/HANISY/AI-Study/blob/main/img/RMS2.png?raw=true'>

- 센터링 과정 생략이 성능에 큰 영향을 주지 않으면서 계산 속도를 높임
- 어텐션 층과 피드포워드 네트워크 이전에 정규화를 적용하여 깊은 모델의 학습 안정성 확보

**위치 인코딩: 로터리 위치 임베딩 (RoPE)**
- 기존의 절대 위치 임베딩(Absolute PE)과 달리, 쿼리와 키 벡터를 회전시켜 그 각도 차이로 위치 정보를 표현
- 별도의 위치 임베딩 벡터를 더하는 연산이 없어 계산이 간단하며 긴 시퀀스 처리에 유리함

##### **EXAONE-3.5로 상품 질문에 대한 대답 생성하기**

````python
# AutoTokenizer 클래스의 from_pretrined() 클래스 메서드를 사용해 EXAONE의 토크나이저 로드
from transformers import AutoTokenizer

exaone_tokenizer = AutoTokenizer.from_pretrained(
    "LGAI-EXAONE/EXAONE-3.5-2.4B-Instruct"
)
````

````python
# pipeline () 함수의 tokenizer 매개변수로 exaone_tokenizer를 전달
from transformers import pipeline

pipe = pipeline(task="text-generation",
                model="LGAI-EXAONE/EXAONE-3.5-2.4B-Instruct",
                tokenizer=exaone_tokenizer,
                device=0, trust_remote_code=True)
````

````python
# 파이프라인 객체 호출
pipe(messages, max_new_tokens=200)
````

##### **토큰 디코딩 전략**

**디코더 출력값과 로짓(Logit)의 정의**
- 어휘 사전에 포함된 각 토큰에 대해 디코더가 산출하는 점수
- 소프트맥스 함수 적용 전의 원시 점수(Raw Score)인 로짓(Logit)
- 로짓에 소프트맥스 함수 적용 시 확률 분포로 변환 가능

**토큰 선택 주체로서의 딥러닝 프레임워크**
- 출력된 로짓 중 실제 토큰을 선택하는 과정은 모델 자체가 아닌 transformers와 같은 프레임워크의 역할
- 프레임워크별 내부 구현 방식의 미세한 차이 존재
- 동일한 모델, 프롬프트, 옵션을 사용하더라도 프레임워크에 따라 상이한 텍스트 출력 가능성

**그리디 서치(Greedy Search)의 개념 및 특성**
- transformers 패키지의 기본 설정 `do_sample=False`
- 매 단계마다 가장 높은 확률(점수)을 가진 토큰 하나만을 선택하는 방식
- 입력(프롬프트)이 동일할 경우 항상 같은 결과를 반환하는 결정론적 특성 보유

**샘플링 전략으로의 전환**
- `do_sample` 매개변수를 `True`로 설정 시 활성화
- 확률 분포에 기반하여 다음 토큰을 무작위로 선택하는 방식 적용

````python
# 기본 샘플링
# 다섯 개의 토큰에 대한 로짓 가정
import numpy as np

logits = np.array([1, 2, 3, 4, 100])
````

````python
# 배열을 소프트맥스 함수에 통과
from scipy.special import softmax

probas = softmax(logits)
print(probas)
````

마지막 원소의 확률이 거의 1에 가깝고 나머지 원소의 확률은 0에 가까움

이런 확률 분포에서 하나의 토큰을 랜덤하게 선택하면 거의 마지막 원소가 선택될 가능성이 높음

````python
# numpy의 random.multinomial() 함수를 사용해 주어진 확률 분포를 바탕으로 샘플링 실험
np.random.multinomial(100, probas)
````

100번 모두 마지막 원소를 선택된 것을 확인

그리디 서치를 적용하는 것과 별반 다르지 않기 때문에 소프트맥스 함수가 만드는 확률 분포를 조금 변형시켜 주는 기법을 종종 사용

````python
# logits을 100으로 나눠 다시 실행
probas = softmax(logits/100)
np.random.multinomial(100, probas)
````

여전히 마지막 원소가 선택되는 가능성이 높지만, 이전과 달리 처음 네 개의 원소도 어느 정도 선택

**온도 파라미터의 원리 및 효과**
- 로짓 스케일링을 통한 확률 분포 변환
  - 소프트맥스 함수 적용 전 로짓(Logit) 값을 온도 값($T$)으로 나누어 스케일링을 수행함
  - $T > 1$ (고온): 로짓 간의 차이가 줄어들어 확률 분포가 평탄(Flat)해짐. 확률이 낮은 토큰의 선택 가능성이 높아져 생성 결과의 다양성과 무작위성이 증가함
  - $T < 1$ (저온): 로짓 간의 차이가 증폭되어 확률 분포가 뾰족(Sharp)해짐. 확률이 높은 토큰에 가중치가 쏠려 그리디 서치와 유사한 결정론적이고 보수적인 결과가 생성됨
- 다양성 조절 역할
  - 생성되는 텍스트의 창의성과 일관성 사이의 균형을 조절하는 핵심 하이퍼파라미터
  - 값이 높을수록 창의적이나 엉뚱한 결과, 낮을수록 정확하고 예측 가능한 결과 유도
- 파이프라인에서의 활용
  - transformers 라이브러리의 파이프라인 객체 또는 생성 함수(generate)에서 temperature 매개변수를 통해 제어 가능
  - 수학적인 기본값은 1.0이며, 이 경우 로짓 값에 변화를 주지 않음

````python
# 온도를 높였을 때 어떤 텍스트가 생성되는지 확인
output = pipe(messages, max_new_tokens=200, return_full_text=False,
              do_sample=True, temperature=10.0)
print(output[0]['generated_text'])
````

토큰의 선택 가능성을 고르게 부여하려다 보니 적절하지 않은 토큰이 많이 선택된 것을 확인

````python
# 온도 파라미터를 낮춰 가장 큰 로짓을 가진 토큰에 높은 가능성 부여하기
output = pipe(messages, max_new_tokens=200, return_full_text=False,
              do_sample=True, temperature=0.001)
print(output[0]['generated_text'])
````

온도 파라미터를 1보다 작게 하니 앞서 `do_sample=False`로 생성한 결과와 동일함 = 그리디 서치를 수행한 것과 다름 없음

transformers 패키지는 이렇게 온도 파라미터만으로도 출력의 결과에 영향을 줄 수 있음

**top-k 샘플링**

- 모델이 출력한 로짓 값을 기준으로 확률이 가장 높은 상위 $k$개의 토큰만을 후보로 추출하는 기법
- 전체 어휘 사전이 아닌, 가능성이 검증된 소수의 후보군으로 선택 범위를 제한함
- 선택된 $k$개의 토큰에 대해서만 소프트맥스 함수를 다시 적용하여 확률 분포를 재정규화 수행<br>
재조정된 확률 분포 내에서 무작위 샘플링을 통해 최종적으로 하나의 토큰을 선택함
- `top_k` 매개변수를 사용하여 후보군의 크기 제어<br>
1보다 큰 정수를 지정해야 하며, 일반적으로 5~50 사이의 값을 많이 사용

````python
output = pipe(messages, max_new_tokens=200, return_full_text=False,
              do_sample=True, top_k=100)
print(output[0]['generated_text'])
````

transformers 패키지는 온도 파라미터를 top-k 샘플링이나 top-p 샘플링보다 먼저 적용

즉, 모델이 생성한 로짓에 대해 temperature를 적용해 분포를 적용한 다음 top_k 매개변수 값만큼 최상위 k개의 로짓을 선택하고, 마지막으로 소프트맥스 함수가 적용

temperature 값을 조정하면 모델이 생성하는 텍스트의 다양성을 조정할 수 있음

````python
# temperature 매개변수를 10.0으로 변경
output = pipe(messages, max_new_tokens=200, return_full_text=False,
              do_sample=True, top_k=10, temperature=10.0)
print(output[0]['generated_text'])
````

맞춤법이 틀리긴 했지만 어느정도 이해할 수 있는 문장이 생성됨

temperature 매개변수로 로짓이 작은 토큰들의 선택 가능성을 높였지만, top_k 매개변수를 함께 사용해 최상위 토큰만 선택한 결과로 볼 수 있음

**top-p 샘플링**

- 뉴클리어스(Nucleus) 샘플링이라고도 불리며, 확률 분포의 모양에 따라 후보군의 크기를 동적으로 조절하는 기법
- Top-k의 경직된 후보 선정 방식을 보완하기 위해 고안
- 확률이 높은 순서대로 토큰을 정렬한 후, 상위 토큰들의 누적 확률 합계가 지정된 top_p 값에 도달할 때까지의 토큰들만 후보로 선택
- `top_p` 매개변수에 0.0 초과 1.0 미만의 실숫값을 지정하여 제어



````python
output = pipe(messages, max_new_tokens=200, return_full_text=False,
              do_sample=True, top_p=0.9)
print(output[0]['generated_text'])
````

`top_p = 0.9`로 지정했다는 것은 LLM이 다음 토큰을 선택할 때 모두 합쳐서 90% 확률 안에 드는 토큰만을 사용했단느 의미

top-k 방식은 최상위 토큰 개수를 고정하기 때문에 비교적 높은 확률을 가진 토큰임에도 불구하고 선택에서 제외될 가능성이 있음

이에 반해 top-p 샘플링은 대상 토큰을 누적 확률로 지정하기 때문에 다양한 확률 분포에 유연하게 대처 가능

top-p 샘플링 순서

##### **오픈AI API로 상품 질문에 대한 대답 생성하기**

````python
from openai import OpenAI
````

````python
# 클라이언트 객체 생성
client = OpenAI(api_key="")
````

**채팅완성 API**
- OpenAI API 사용의 가장 기본이자 핵심이 되는 인터페이스
- 단발성 텍스트 완성을 넘어, 대화의 문맥을 고려하여 응답을 생성하는 기능 수행
- OpenAI 클라이언트 객체(예: client의 `chat.completions.create()` 메서드를 사용하여 API 호출
- model: 호출하여 사용할 구체적인 모델의 ID를 문자열로 지정 (예: 'gpt-4o', 'gpt-3.5-turbo')
- messages: 모델과 주고받을 대화 내역을 담은 리스트 입력
-각 메시지는 역할(Role)과 내용(Content)으로 구성된 딕셔너리 객체의 배열 형태

````python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages
)
````

````python
print(completion.choices[0].message.content)
````

OpenAI는 top-p 샘플링을 위한 `top_p` 매개변수를 제공하고 있음

`top_p` 매개변수의 기본값은 1로 전체 토큰을 사용

이를 0.9 정도로 낮추면 높은 확률을 가진 토큰이 선택될 가능성이 조금 더 올라감

````python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
    top_p=0.9
)
print(completion.choices[0].message.content)
````

OpenAI의 `temperature` 매개변수는 0~2 사이의 값을 지정해야 하며 기본값은 1

````python
# 온도파라미터를 조정하여 랜덤한 출력 생성
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
    temperature=1.8
)
print(completion.choices[0].message.content)
````

temperature 값을 증가시키니 불안정한 답 출력

OpenAI에서는 temperature 매개변수와 top_p 매개변수를 동시에 사용하는 것보다 둘 중 하나를 선택해서 사용하도록 권장

OpenAI API에서 디코딩 전략의 우선 순위나 그리디 서치 방식을 사용하는 방법은 명확하게 문서화되어 있지 않음

temperature=0으로 지정하면 일정한 출력을 얻을 수 있다고 알려져 있지만 예고 없이 언제든 동작이 바뀔 수 있음
